```python
import pandas as pd
import numpy as np
import re
df = pd.read_csv("Bangladesh_database_Final_Merged.csv")

df.head()
```

```python
print("Kieu du lieu cua tung cot")
print(df.dtypes)
print()
print("Thong ke mo ta cac cot so")
df.describe()
```

```python
print("Thong ke cac cot phan loai")
cat_cols = ['District', 'Dominant_Soil_Texture', 'AP Ratio',
            'Growth', 'Crop Name', 'Transplant', 'Season']
 
for col in cat_cols:
    print(f" [{col}] — {df[col].nunique()} gia tri duy nhat")
    print(df[col].value_counts().head(5).to_string())
```

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
 
missing_df = pd.DataFrame({
    'Số lượng NaN': missing,
    'Tỷ lệ (%)': missing_pct
})
 
print("So gia tri thieu theo cot")
print(missing_df[missing_df['Số lượng NaN'] > 0])
 
if missing.sum() == 0:
    print("\nKhong co gia tri thieu trong du lieu")
```

```python
n_dup = df.duplicated().sum()
print(f"So dong trung lap: {n_dup}")
 
if n_dup > 0:
    print("Cac dong bi trung:")
    print(df[df.duplicated(keep=False)])
    df = df.drop_duplicates()
    
else:
    print("Khong co dong trung lap")

# Kiem tra va loai cac dong Crop Name bi loi #REF!
df = df[df['Crop Name'] != '#REF!']
print(f"Kich thuoc sau khi xoa dong #REF!: {df.shape}")
```

```python
print(f"Kieu du lieu AP Ratio ban dau: {df['AP Ratio'].dtype}")  

# Kiem tra cac gia tri AP Ratio khong the chuyen sang so
failed_mask = pd.to_numeric(df['AP Ratio'], errors='coerce').isnull()
print(f"So gia tri AP Ratio khong hop le: {failed_mask.sum()}")
print("Cac gia tri loi:")
print(df[failed_mask]['AP Ratio'].value_counts())

# Cac loi #DIV/0! se thanh NaN khi chuyen sang kieu so

# Chuyen AP Ratio sang kieu so
df['AP Ratio'] = pd.to_numeric(df['AP Ratio'], errors='coerce')
 
# Dien gia tri thieu theo trung vi cua tung loai cay
# Moi loai cay co mot muc AP Ratio trung vi rieng
crop_yield_imputed = df.groupby('Crop Name')['AP Ratio'].transform('median')
df['AP Ratio'] = df['AP Ratio'].fillna(crop_yield_imputed)

# Dung trung vi toan bo cot neu van con gia tri thieu
df['AP Ratio'] = df['AP Ratio'].fillna(df['AP Ratio'].median())
 
print("\nAP Ratio sau khi xu ly")
print(f"Kieu du lieu moi: {df['AP Ratio'].dtype}")
print(f"So gia tri thieu con lai: {df['AP Ratio'].isnull().sum()}")
```

```python
print("Truoc khi xu ly")
print("Cac ten cot co khoang trang:")
for col in df.columns:
    if ' ' in col:
        print(f"    '{col}'")
 # Ten cot co khoang trang khong tien khi truy cap bang thuoc tinh


df.columns = df.columns.str.strip().str.replace(' ', '_')
 
print("Sau khi xu ly")
print("Ten cot sau khi chuan hoa:")
print(list(df.columns))
```

```python
print(df['Dominant_Soil_Texture'].value_counts())
# Bo phan chu thich tieng Viet trong ten loai dat
 
df['Dominant_Soil_Texture'] = df['Dominant_Soil_Texture'].apply(
   lambda x: re.sub(r'\s*\([^)]*\)', '', str(x)).strip()
)
 
print("Sau khi xu ly")
print(df['Dominant_Soil_Texture'].value_counts())
```

```python
# Kiem tra khoang trang thua trong Growth
double_space = df['Growth'].str.contains('  ', na=False)
print("Gia tri co khoang trang thua:", df[double_space]['Growth'].value_counts().to_dict())
 
# Kiem tra cach viet hoa, viet thuong
bad_case = df['Growth'].str.contains('[a-z]o [a-z]', na=False)  # "to july"
print("\nGia tri viet hoa, viet thuong chua dong nhat:")
print(df[bad_case]['Growth'].value_counts().to_dict())
# Chuan hoa khoang trang va cach viet Growth

# Xoa khoang trang thua
df['Growth'] = df['Growth'].str.split().str.join(' ')
# Dua ve Title Case
df['Growth'] = df['Growth'].str.title()
 
print("Sau khi xu ly")
print(df['Growth'].value_counts())
```

```python

print("Cac gia tri:")
print(sorted(df['Transplant'].unique()))
print(f"\nSo cach viet thang ban dau: {df['Transplant'].nunique()}")
# Quy cac ten thang viet tat ve ten day du
 
month_map = {
    'Jan': 'January',   'Feb': 'February', 'Mar': 'March',
    'Apr': 'April',     'Jun': 'June',     'Jul': 'July',
    'Aug': 'August',    'Sep': 'September','Oct': 'October',
    'Nov': 'November',  'Dec': 'December'
}
df['Transplant'] = df['Transplant'].replace(month_map)
 

print("Cac gia tri:")
print(sorted(df['Transplant'].unique()))
```

```python
n_neg = (df['NDVI'] < 0).sum()
print("Truoc khi xu ly")
print(f"So gia tri NDVI am: {n_neg}")
print(df[df['NDVI'] < 0]['NDVI'].describe())
# Cac gia tri NDVI am trong tap nay duoc xem la nhieu cam bien

# Dua cac gia tri NDVI am ve 0
df['NDVI'] = df['NDVI'].clip(lower=0)
print("Sau khi xu ly")
print(f"So gia tri NDVI am con lai: {(df['NDVI'] < 0).sum()}")
print(f"NDVI min: {df['NDVI'].min():.6f}")
```

```python
# Dam bao cac cot do luong co kieu so
numeric_fix_cols = ['Avg Temp', 'Avg Humidity', 'Max Temp', 'Min Temp', 'Rainfall', 'Wind_Mean', 'pH']
for col in numeric_fix_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        df[col] = df[col].fillna(df[col].mean())


# Chuyen LST tu Kelvin sang do C
if 'LST_Kelvin' in df.columns:
    df['LST_C'] = df['LST_Kelvin'] - 273.15
    df = df.drop(columns=['LST_Kelvin'])



# Sua thu tu cot do am neu bi dao
if 'Max Relative Humidity' in df.columns and 'Min Relative Humidity' in df.columns:
    # Dam bao do am lon nhat khong nho hon do am nho nhat
    mask = df['Max Relative Humidity'] < df['Min Relative Humidity']
    df.loc[mask, ['Min Relative Humidity', 'Max Relative Humidity']] = df.loc[mask, ['Max Relative Humidity', 'Min Relative Humidity']].values

```

```python
output_file = 'Agri_Data_Cleaned.csv'
df.to_csv(output_file, index=False, encoding='utf-8-sig')
print(f"Da luu du lieu sach tai: {output_file}")
```
